```python
# ============================================================
# PlaceMux Task 24 - Python Data Analysis
# Disaster Recovery, Chaos Testing & Business Continuity
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. Upload Cleaned Excel File
# ------------------------------------------------------------

from google.colab import files

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_excel(file_name)

print("File loaded successfully")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])


# ------------------------------------------------------------
# 2. Basic Dataset Information
# ------------------------------------------------------------

print("\n========== DATASET INFORMATION ==========")

print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())


# ------------------------------------------------------------
# 3. Basic Statistical Analysis
# ------------------------------------------------------------

print("\n========== STATISTICAL SUMMARY ==========")

display(df.describe())


# ------------------------------------------------------------
# 4. Total Records
# ------------------------------------------------------------

total_records = len(df)

print("\nTotal Records:", total_records)


# ------------------------------------------------------------
# 5. Metric Analysis
# ------------------------------------------------------------

if "metric_value" in df.columns:

    total_metric_value = df["metric_value"].sum()
    average_metric_value = df["metric_value"].mean()
    maximum_metric_value = df["metric_value"].max()
    minimum_metric_value = df["metric_value"].min()

    print("\n========== METRIC ANALYSIS ==========")

    print("Total Metric Value:", total_metric_value)
    print("Average Metric Value:", round(average_metric_value, 2))
    print("Maximum Metric Value:", maximum_metric_value)
    print("Minimum Metric Value:", minimum_metric_value)


# ------------------------------------------------------------
# 6. Source System Analysis
# ------------------------------------------------------------

if "source_system" in df.columns:

    source_analysis = (
        df.groupby("source_system")
        .agg(
            Records=("event_id", "count"),
            Total_Metric_Value=("metric_value", "sum"),
            Average_Metric_Value=("metric_value", "mean")
        )
        .reset_index()
    )

    source_analysis["Average_Metric_Value"] = (
        source_analysis["Average_Metric_Value"].round(2)
    )

    print("\n========== SOURCE SYSTEM ANALYSIS ==========")

    display(source_analysis)


# ------------------------------------------------------------
# 7. Metric Type Analysis
# ------------------------------------------------------------

if "metric_name" in df.columns:

    metric_analysis = (
        df.groupby("metric_name")
        .agg(
            Records=("event_id", "count"),
            Total_Value=("metric_value", "sum"),
            Average_Value=("metric_value", "mean")
        )
        .reset_index()
    )

    metric_analysis["Average_Value"] = (
        metric_analysis["Average_Value"].round(2)
    )

    print("\n========== METRIC TYPE ANALYSIS ==========")

    display(metric_analysis)


# ------------------------------------------------------------
# 8. Backup Status Analysis
# ------------------------------------------------------------

if "backup_status" in df.columns:

    backup_analysis = (
        df["backup_status"]
        .value_counts()
        .reset_index()
    )

    backup_analysis.columns = [
        "Backup_Status",
        "Records"
    ]

    backup_analysis["Percentage"] = (
        backup_analysis["Records"] /
        backup_analysis["Records"].sum() * 100
    ).round(2)

    print("\n========== BACKUP STATUS ANALYSIS ==========")

    display(backup_analysis)


# ------------------------------------------------------------
# 9. Restore Status Analysis
# ------------------------------------------------------------

if "restore_status" in df.columns:

    restore_analysis = (
        df["restore_status"]
        .value_counts()
        .reset_index()
    )

    restore_analysis.columns = [
        "Restore_Status",
        "Records"
    ]

    restore_analysis["Percentage"] = (
        restore_analysis["Records"] /
        restore_analysis["Records"].sum() * 100
    ).round(2)

    print("\n========== RESTORE STATUS ANALYSIS ==========")

    display(restore_analysis)


# ------------------------------------------------------------
# 10. Backfill Status Analysis
# ------------------------------------------------------------

if "backfill_status" in df.columns:

    backfill_analysis = (
        df["backfill_status"]
        .value_counts()
        .reset_index()
    )

    backfill_analysis.columns = [
        "Backfill_Status",
        "Records"
    ]

    backfill_analysis["Percentage"] = (
        backfill_analysis["Records"] /
        backfill_analysis["Records"].sum() * 100
    ).round(2)

    print("\n========== BACKFILL STATUS ANALYSIS ==========")

    display(backfill_analysis)


# ------------------------------------------------------------
# 11. Reporting Status Analysis
# ------------------------------------------------------------

if "reporting_status" in df.columns:

    reporting_analysis = (
        df["reporting_status"]
        .value_counts()
        .reset_index()
    )

    reporting_analysis.columns = [
        "Reporting_Status",
        "Records"
    ]

    reporting_analysis["Percentage"] = (
        reporting_analysis["Records"] /
        reporting_analysis["Records"].sum() * 100
    ).round(2)

    print("\n========== REPORTING STATUS ANALYSIS ==========")

    display(reporting_analysis)


# ------------------------------------------------------------
# 12. RTO Analysis
# ------------------------------------------------------------

if "rto_minutes" in df.columns:

    average_rto = df["rto_minutes"].mean()
    minimum_rto = df["rto_minutes"].min()
    maximum_rto = df["rto_minutes"].max()

    print("\n========== RTO ANALYSIS ==========")

    print("Average RTO:", round(average_rto, 2), "minutes")
    print("Minimum RTO:", minimum_rto, "minutes")
    print("Maximum RTO:", maximum_rto, "minutes")


# ------------------------------------------------------------
# 13. RPO Analysis
# ------------------------------------------------------------

if "rpo_minutes" in df.columns:

    average_rpo = df["rpo_minutes"].mean()
    minimum_rpo = df["rpo_minutes"].min()
    maximum_rpo = df["rpo_minutes"].max()

    print("\n========== RPO ANALYSIS ==========")

    print("Average RPO:", round(average_rpo, 2), "minutes")
    print("Minimum RPO:", minimum_rpo, "minutes")
    print("Maximum RPO:", maximum_rpo, "minutes")


# ------------------------------------------------------------
# 14. RTO Compliance Analysis
# ------------------------------------------------------------

if "rto_compliance" in df.columns:

    rto_compliance = (
        df["rto_compliance"]
        .value_counts()
        .reset_index()
    )

    rto_compliance.columns = [
        "RTO_Compliance",
        "Records"
    ]

    rto_compliance["Percentage"] = (
        rto_compliance["Records"] /
        rto_compliance["Records"].sum() * 100
    ).round(2)

    print("\n========== RTO COMPLIANCE ==========")

    display(rto_compliance)


# ------------------------------------------------------------
# 15. RPO Compliance Analysis
# ------------------------------------------------------------

if "rpo_compliance" in df.columns:

    rpo_compliance = (
        df["rpo_compliance"]
        .value_counts()
        .reset_index()
    )

    rpo_compliance.columns = [
        "RPO_Compliance",
        "Records"
    ]

    rpo_compliance["Percentage"] = (
        rpo_compliance["Records"] /
        rpo_compliance["Records"].sum() * 100
    ).round(2)

    print("\n========== RPO COMPLIANCE ==========")

    display(rpo_compliance)


# ------------------------------------------------------------
# 16. Overall Recovery Analysis
# ------------------------------------------------------------

if "overall_recovery_status" in df.columns:

    recovery_analysis = (
        df["overall_recovery_status"]
        .value_counts()
        .reset_index()
    )

    recovery_analysis.columns = [
        "Recovery_Status",
        "Records"
    ]

    recovery_analysis["Percentage"] = (
        recovery_analysis["Records"] /
        recovery_analysis["Records"].sum() * 100
    ).round(2)

    print("\n========== OVERALL RECOVERY ANALYSIS ==========")

    display(recovery_analysis)


# ------------------------------------------------------------
# 17. Reporting Continuity Analysis
# ------------------------------------------------------------

if "reporting_continuity" in df.columns:

    continuity_analysis = (
        df["reporting_continuity"]
        .value_counts()
        .reset_index()
    )

    continuity_analysis.columns = [
        "Reporting_Continuity",
        "Records"
    ]

    continuity_analysis["Percentage"] = (
        continuity_analysis["Records"] /
        continuity_analysis["Records"].sum() * 100
    ).round(2)

    print("\n========== REPORTING CONTINUITY ==========")

    display(continuity_analysis)


# ------------------------------------------------------------
# 18. Outage Analysis
# ------------------------------------------------------------

if "outage_minutes" in df.columns:

    average_outage = df["outage_minutes"].mean()
    maximum_outage = df["outage_minutes"].max()
    minimum_outage = df["outage_minutes"].min()

    print("\n========== OUTAGE ANALYSIS ==========")

    print("Average Outage:", round(average_outage, 2), "minutes")
    print("Minimum Outage:", minimum_outage, "minutes")
    print("Maximum Outage:", maximum_outage, "minutes")


# ------------------------------------------------------------
# 19. Source System + Recovery Analysis
# ------------------------------------------------------------

if (
    "source_system" in df.columns
    and "overall_recovery_status" in df.columns
):

    source_recovery = pd.crosstab(
        df["source_system"],
        df["overall_recovery_status"]
    ).reset_index()

    print("\n========== SOURCE SYSTEM RECOVERY ==========")

    display(source_recovery)


# ------------------------------------------------------------
# 20. Source System + Reporting Continuity
# ------------------------------------------------------------

if (
    "source_system" in df.columns
    and "reporting_continuity" in df.columns
):

    source_continuity = pd.crosstab(
        df["source_system"],
        df["reporting_continuity"]
    ).reset_index()

    print("\n========== SOURCE SYSTEM REPORTING CONTINUITY ==========")

    display(source_continuity)


# ------------------------------------------------------------
# 21. Daily Analysis
# ------------------------------------------------------------

if "event_date" in df.columns:

    daily_analysis = (
        df.groupby("event_date")
        .agg(
            Records=("event_id", "count"),
            Total_Metric_Value=("metric_value", "sum"),
            Average_RTO=("rto_minutes", "mean"),
            Average_RPO=("rpo_minutes", "mean"),
            Average_Outage=("outage_minutes", "mean")
        )
        .reset_index()
    )

    daily_analysis["Average_RTO"] = (
        daily_analysis["Average_RTO"].round(2)
    )

    daily_analysis["Average_RPO"] = (
        daily_analysis["Average_RPO"].round(2)
    )

    daily_analysis["Average_Outage"] = (
        daily_analysis["Average_Outage"].round(2)
    )

    print("\n========== DAILY ANALYSIS ==========")

    display(daily_analysis.head(20))


# ------------------------------------------------------------
# 22. Create KPI Summary
# ------------------------------------------------------------

kpi_data = {
    "KPI": [
        "Total Records",
        "Total Metric Value",
        "Average Metric Value",
        "Average RTO (Minutes)",
        "Average RPO (Minutes)",
        "Average Outage (Minutes)"
    ],
    "Value": [
        len(df),
        df["metric_value"].sum(),
        round(df["metric_value"].mean(), 2),
        round(df["rto_minutes"].mean(), 2),
        round(df["rpo_minutes"].mean(), 2),
        round(df["outage_minutes"].mean(), 2)
    ]
}

kpi_summary = pd.DataFrame(kpi_data)

print("\n========== KPI SUMMARY ==========")

display(kpi_summary)


# ------------------------------------------------------------
# 23. Export All Analysis Tables to Excel
# ------------------------------------------------------------

output_file = "PlaceMux_Task24_Python_Analysis.xlsx"

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:

    df.to_excel(
        writer,
        sheet_name="Cleaned_Data",
        index=False
    )

    kpi_summary.to_excel(
        writer,
        sheet_name="KPI_Summary",
        index=False
    )

    if "source_analysis" in locals():
        source_analysis.to_excel(
            writer,
            sheet_name="Source_Analysis",
            index=False
        )

    if "metric_analysis" in locals():
        metric_analysis.to_excel(
            writer,
            sheet_name="Metric_Analysis",
            index=False
        )

    if "backup_analysis" in locals():
        backup_analysis.to_excel(
            writer,
            sheet_name="Backup_Analysis",
            index=False
        )

    if "restore_analysis" in locals():
        restore_analysis.to_excel(
            writer,
            sheet_name="Restore_Analysis",
            index=False
        )

    if "backfill_analysis" in locals():
        backfill_analysis.to_excel(
            writer,
            sheet_name="Backfill_Analysis",
            index=False
        )

    if "reporting_analysis" in locals():
        reporting_analysis.to_excel(
            writer,
            sheet_name="Reporting_Analysis",
            index=False
        )

    if "rto_compliance" in locals():
        rto_compliance.to_excel(
            writer,
            sheet_name="RTO_Compliance",
            index=False
        )

    if "rpo_compliance" in locals():
        rpo_compliance.to_excel(
            writer,
            sheet_name="RPO_Compliance",
            index=False
        )

    if "recovery_analysis" in locals():
        recovery_analysis.to_excel(
            writer,
            sheet_name="Recovery_Analysis",
            index=False
        )

    if "continuity_analysis" in locals():
        continuity_analysis.to_excel(
            writer,
            sheet_name="Continuity_Analysis",
            index=False
        )

    if "source_recovery" in locals():
        source_recovery.to_excel(
            writer,
            sheet_name="Source_Recovery",
            index=False
        )

    if "source_continuity" in locals():
        source_continuity.to_excel(
            writer,
            sheet_name="Source_Continuity",
            index=False
        )

    if "daily_analysis" in locals():
        daily_analysis.to_excel(
            writer,
            sheet_name="Daily_Analysis",
            index=False
        )


print("\n======================================")
print("ANALYSIS COMPLETED SUCCESSFULLY")
print("======================================")

print("\nAnalysis file:", output_file)


# ------------------------------------------------------------
# 24. Download Analysis File
# ------------------------------------------------------------

files.download(output_file)
```
